## The General Causal Assumption is that:

- behavioral_outside_home -> behavioral_large_gatherings -> behavioral_face_mask -> behavioral_wash_hands -> behavioral_touch_face -> chronic_med_condition

- hhs_geo_region -> behavioral_large_gatherings -> behavioral_touch_face -> chronic_med_condition (can be a genuine causation or bilateral causation)

-  employment_occupation -> income_poverty -> doctor_recc_h1h1

in africa, majority of these happen due to the fact that males are often the ones that find jobs and it was an economic phenomenon that was discovered by economists. For the 2nd causal assumption, it depends on the regions and its traditions of the spread of the h1h1 virus that leads to chronic_mde_condition.

## Feature Engineering

```python
def feature_engineer(df):

    df = df.copy()



    linear_cols = [
        "behavioral_outside_home",
        "behavioral_large_gatherings",
        "behavioral_face_mask",
        "behavioral_wash_hands",
        "behavioral_touch_face"
    ]

    X = df[linear_cols].fillna(0)

    pca = PCA(n_components=2)

    pca_features = pca.fit_transform(X)

    df["behavior_linear1"] = pca_features[:,0]
    df["behavior_linear2"] = pca_features[:,1]



    # total risk behavior

    df["risk_behavior_sum"] = (
        df["behavioral_outside_home"] +
        df["behavioral_large_gatherings"] +
        (1-df["behavioral_face_mask"]) +
        (1-df["behavioral_wash_hands"]) +
        df["behavioral_touch_face"]
    )


    # hygiene score

    df["hygiene_score"] = (
        df["behavioral_face_mask"] +
        df["behavioral_wash_hands"]
    )


    # risk ratio

    df["risk_hygiene_ratio"] = (
        df["risk_behavior_sum"] /
        (df["hygiene_score"]+1)
    )



    # combine behavior + health

    df["behavior_health_interaction"] = (
        df["risk_behavior_sum"] *
        df["chronic_med_condition"]
    )


    # behavior × geography

    df["region_behavior"] = (
        df["hhs_geo_region"] *
        df["behavioral_large_gatherings"]
    )


    # occupation × poverty

    df["occupation_poverty"] = (
        df["employment_occupation"] *
        df["income_poverty"]
    )



    # healthcare accessibility proxy

    df["healthcare_access"] = (
        df["income_poverty"] *
        (1+df["doctor_recc_h1n1"])
    )


    # socioeconomic pressure

    df["economic_pressure"] = (
        df["employment_occupation"] +
        df["income_poverty"]
    )




    # cooperation score

    # assumes masks + washing hands
    # benefit everyone

    df["cooperation_score"] = (
        df["behavioral_face_mask"] +
        df["behavioral_wash_hands"]
    )


    # defection score

    df["defection_score"] = (
        df["behavioral_large_gatherings"] +
        df["behavioral_touch_face"]
    )


    # payoff approximation

    df["social_payoff"] = (
        df["cooperation_score"] -
        df["defection_score"]
    )


    # CAUSAL PATH FEATURES


    # behavioral chain propagation

    df["behavior_chain"] = (
        df["behavioral_outside_home"] *
        df["behavioral_large_gatherings"] *
        (1+df["behavioral_touch_face"])
    )


    # regional pathway

    df["region_chain"] = (
        df["hhs_geo_region"] *
        df["behavioral_large_gatherings"] *
        (1+df["behavioral_touch_face"])
    )


    # economic pathway

    df["economic_chain"] = (
        df["employment_occupation"] *
        df["income_poverty"] *
        (1+df["doctor_recc_h1n1"])
    )


    df.replace([np.inf,-np.inf],0,inplace=True)

    return df
```

```python
train = feature_engineer(train)
test = feature_engineer(test)
```

## Modelling

```python
# For H1N1 Vaccine
h1n1_train = train.copy()
X_h1n1 = h1n1_train.drop(columns=["h1n1_vaccine", "respondent_id"], axis=1)
y_h1n1 = h1n1_train["h1n1_vaccine"]

X_h1n1_train, X_h1n1_test, y_h1n1_train, y_h1n1_test = train_test_split(X_h1n1, y_h1n1, stratify=y_h1n1, test_size=0.2, random_state=42)

#For Seasonal Vaccine
seasonal_train = train.copy()
X_seasonal = seasonal_train.drop(columns=["seasonal_vaccine", "respondent_id"], axis=1)
y_seasonal = seasonal_train["seasonal_vaccine"]

X_seasonal_train, X_seasonal_test, y_seasonal_train, y_seasonal_test = train_test_split(X_seasonal, y_seasonal, stratify=y_seasonal, test_size=0.2, random_state=42)
```

```python
train
```

```python
lr = Pipeline([
    ("scaler", RobustScaler()),
    ("model", LogisticRegression(
        penalty="l2",
        C=0.5,
        solver="saga",
        max_iter=5000,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])


rdg = Pipeline([
    ("scaler", RobustScaler()),
    ("model", RidgeClassifier(
        alpha=3.0,
        class_weight="balanced",
        random_state=42
    ))
])


sgd = Pipeline([
    ("scaler", RobustScaler()),
    ("model", SGDClassifier(
        loss="modified_huber",
        penalty="elasticnet",
        alpha=1e-4,
        l1_ratio=0.15,
        max_iter=3000,
        early_stopping=True,
        class_weight="balanced",
        random_state=42
    ))
])



adaboost = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", AdaBoostClassifier(
        estimator=DecisionTreeClassifier(
            max_depth=2,
            min_samples_leaf=5
        ),
        n_estimators=300,
        learning_rate=0.03,
        random_state=42
    ))
])


rfr = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", RandomForestClassifier(
        n_estimators=500,
        max_depth=12,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features="sqrt",
        class_weight="balanced_subsample",
        n_jobs=-1,
        random_state=42
    ))
])


gbt = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", GradientBoostingClassifier(
        n_estimators=300,
        learning_rate=0.03,
        max_depth=5,
        subsample=0.8,
        min_samples_split=5,
        random_state=42
    ))
])


xgb = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", XGBClassifier(
        n_estimators=1000,
        learning_rate=0.02,
        max_depth=6,
        min_child_weight=3,
        subsample=0.8,
        colsample_bytree=0.8,
        gamma=0.1,
        reg_alpha=0.5,
        reg_lambda=2,
        eval_metric="logloss",
        random_state=42
    ))
])


lgbm = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", LGBMClassifier(
        n_estimators=1000,
        learning_rate=0.02,
        num_leaves=31,
        max_depth=8,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.5,
        reg_lambda=2,
        class_weight="balanced",
        random_state=42,
        verbose=-1
    ))
])


catboost = Pipeline([
    ("transformer",
     QuantileTransformer(
         output_distribution="normal",
         random_state=42
     )),
    
    ("model", CatBoostClassifier(
        iterations=1000,
        learning_rate=0.02,
        depth=6,
        l2_leaf_reg=5,
        loss_function="Logloss",
        eval_metric="F1",
        random_seed=42,
        verbose=0
    ))
])
```

```python
#Mlflow for h1n1_vaccine cure
mlflow.set_experiment("H1N1_Model_Training_and_Experiments")

models = {
    "LogisticRegression": lr,
    "StochasticGradientClassifier": sgd,
    "RidgeClassifier": rdg,
    "AdaBoostClassifier": adaboost,
    "RandomForestClassifier": rfr,
    "GradientBoostingClassifier": gbt,
    "XGBoostClassifier": xgb,
    "LGBMClassifier": lgbm,
    "CatBoostClassifier": catboost
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for name, model in models.items():

    print("="*70)
    print(name)
    print("="*70)

    with mlflow.start_run(run_name=name):

        start = time.time()

        cv_f1 = cross_val_score(
            model,
            X_h1n1_train,
            y_h1n1_train,
            cv=cv,
            scoring="f1",
            n_jobs=-1
        )

        cv_precision = cross_val_score(
            model,
            X_h1n1_train,
            y_h1n1_train,
            cv=cv,
            scoring="precision",
            n_jobs=-1
        )

        cv_recall = cross_val_score(
            model,
            X_h1n1_train,
            y_h1n1_train,
            cv=cv,
            scoring="recall",
            n_jobs=-1
        )


        model.fit(
            X_h1n1_train,
            y_h1n1_train
        )


        y_pred = model.predict(
            X_h1n1_test
        )

        if hasattr(model, "predict_proba"):

            y_proba = model.predict_proba(
                X_h1n1_test
            )[:,1]

        else:

            decision = model.decision_function(
                X_h1n1_test
            )

            y_proba = (
                decision - decision.min()
            ) / (
                decision.max() - decision.min()
            )


        acc = accuracy_score(
            y_h1n1_test,
            y_pred
        )

        f1 = f1_score(
            y_h1n1_test,
            y_pred
        )

        precision = precision_score(
            y_h1n1_test,
            y_pred
        )

        recall = recall_score(
            y_h1n1_test,
            y_pred
        )

        auc = roc_auc_score(
            y_h1n1_test,
            y_proba
        )

        brier = brier_score_loss(
            y_h1n1_test,
            y_proba
        )

        cm = confusion_matrix(
            y_h1n1_test,
            y_pred
        )

        report = classification_report(
            y_h1n1_test,
            y_pred
        )

        runtime = time.time() - start


        print(
            f"CV F1: {cv_f1.mean():.4f} ± {cv_f1.std():.4f}"
        )

        print(
            f"CV Precision: {cv_precision.mean():.4f}"
        )

        print(
            f"CV Recall: {cv_recall.mean():.4f}"
        )

        print(
            f"Accuracy: {acc:.4f}"
        )

        print(
            f"F1: {f1:.4f}"
        )

        print(
            f"Precision: {precision:.4f}"
        )

        print(
            f"Recall: {recall:.4f}"
        )

        print(
            f"ROC-AUC: {auc:.4f}"
        )

        print(
            f"Brier Score: {brier:.4f}"
        )

        print(
            f"Runtime: {runtime:.2f}s"
        )

        print("\nConfusion Matrix:")
        print(cm)

        print("\nClassification Report:")
        print(report)


        fig, ax = plt.subplots()

        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm
        )

        disp.plot(ax=ax)

        plt.savefig(
            "confusion_matrix.png"
        )

        plt.close()



        mlflow.log_metric(
            "cv_f1_mean",
            cv_f1.mean()
        )

        mlflow.log_metric(
            "cv_f1_std",
            cv_f1.std()
        )

        mlflow.log_metric(
            "accuracy",
            acc
        )

        mlflow.log_metric(
            "f1",
            f1
        )

        mlflow.log_metric(
            "precision",
            precision
        )

        mlflow.log_metric(
            "recall",
            recall
        )

        mlflow.log_metric(
            "roc_auc",
            auc
        )

        mlflow.log_metric(
            "brier_score",
            brier
        )

        mlflow.log_metric(
            "runtime",
            runtime
        )

        mlflow.log_artifact(
            "confusion_matrix.png"
        )

        mlflow.sklearn.log_model(
            model,
            name
        )


        results.append({

            "Model":name,
            "CV_F1":cv_f1.mean(),
            "F1":f1,
            "Precision":precision,
            "Recall":recall,
            "AUC":auc,
            "Brier":brier,
            "Accuracy":acc,
            "Runtime":runtime

        })


results_df = (
    pd.DataFrame(results)
    .sort_values(
        "F1",
        ascending=False
    )
)

print("\n")
print("="*70)
print("FINAL RESULTS")
print("="*70)

print(results_df)

results_df.to_csv(
    "h1n1_results.csv",
    index=False
)
```

```python
#Seasonal Vaccine
mlflow.set_experiment("Seasonal_Vaccine_Model_Training_and_Experiments")

vaccine_models = {
    "LogisticRegression": lr,
    "StochasticGradientClassifier": sgd,
    "RidgeClassifier": rdg,
    "AdaBoostClassifier": adaboost,
    "RandomForestClassifier": rfr,
    "GradientBoostingClassifier": gbt,
    "XGBoostClassifier": xgb,
    "LGBMClassifier": lgbm,
    "CatBoostClassifier": catboost
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for name, model in vaccine_models.items():

    print("="*70)
    print(name)
    print("="*70)

    with mlflow.start_run(run_name=name):

        start = time.time()

        cv_f1 = cross_val_score(
            model,
            X_seasonal_train,
            y_seasonal_train,
            cv=cv,
            scoring="f1",
            n_jobs=-1
        )

        cv_precision = cross_val_score(
            model,
            X_seasonal_train,
            y_seasonal_train,
            cv=cv,
            scoring="precision",
            n_jobs=-1
        )

        cv_recall = cross_val_score(
            model,
            X_seasonal_train,
            y_seasonal_train,
            cv=cv,
            scoring="recall",
            n_jobs=-1
        )

        cv_auc = cross_val_score(
            model,
            X_seasonal_train,
            y_seasonal_train,
            cv=cv,
            scoring="roc_auc",
            n_jobs=-1
        )


        model.fit(
            X_seasonal_train,
            y_seasonal_train
        )


        y_pred = model.predict(
            X_seasonal_test
        )

        if hasattr(model, "predict_proba"):

            y_proba = model.predict_proba(
                X_seasonal_test
            )[:, 1]

        else:

            scores = model.decision_function(
                X_seasonal_test
            )

            y_proba = (
                scores - scores.min()
            ) / (
                scores.max() - scores.min()
            )


        acc = accuracy_score(y_seasonal_test, y_pred)
        f1 = f1_score(y_seasonal_test, y_pred)
        precision = precision_score(y_seasonal_test, y_pred)
        recall = recall_score(y_seasonal_test, y_pred)

        auc = roc_auc_score(y_seasonal_test, y_proba)

        brier = brier_score_loss(y_seasonal_test, y_proba)

        cm = confusion_matrix(y_seasonal_test, y_pred)

        report = classification_report(y_seasonal_test, y_pred)

        runtime = time.time() - start


        print(f"CV F1: {cv_f1.mean():.4f} ± {cv_f1.std():.4f}")
        print(f"CV Precision: {cv_precision.mean():.4f}")
        print(f"CV Recall: {cv_recall.mean():.4f}")
        print(f"CV AUC: {cv_auc.mean():.4f}")

        print(f"Accuracy: {acc:.4f}")
        print(f"F1: {f1:.4f}")
        print(f"Precision: {precision:.4f}")
        print(f"Recall: {recall:.4f}")
        print(f"ROC AUC: {auc:.4f}")
        print(f"Brier Score: {brier:.4f}")
        print(f"Runtime: {runtime:.2f}s")

        print("\nConfusion Matrix:")
        print(cm)

        print("\nClassification Report:")
        print(report)


        fig, ax = plt.subplots()

        disp = ConfusionMatrixDisplay(confusion_matrix=cm)
        disp.plot(ax=ax)

        cm_file = f"{name}_seasonal_cm.png"
        plt.savefig(cm_file)
        plt.close()


        mlflow.log_metric("cv_f1_mean", cv_f1.mean())
        mlflow.log_metric("cv_f1_std", cv_f1.std())
        mlflow.log_metric("cv_precision", cv_precision.mean())
        mlflow.log_metric("cv_recall", cv_recall.mean())
        mlflow.log_metric("cv_auc", cv_auc.mean())

        mlflow.log_metric("accuracy", acc)
        mlflow.log_metric("f1", f1)
        mlflow.log_metric("precision", precision)
        mlflow.log_metric("recall", recall)
        mlflow.log_metric("roc_auc", auc)
        mlflow.log_metric("brier_score", brier)
        mlflow.log_metric("runtime", runtime)

        mlflow.log_artifact(cm_file)

        mlflow.sklearn.log_model(model, name)



        results.append({
            "Model": name,
            "CV_F1": cv_f1.mean(),
            "F1": f1,
            "Precision": precision,
            "Recall": recall,
            "AUC": auc,
            "Brier": brier,
            "Accuracy": acc,
            "Runtime": runtime
        })




seasonal_results_df = pd.DataFrame(results).sort_values(
    "F1",
    ascending=False
)

print("\n")
print("="*70)
print("FINAL SEASONAL RESULTS")
print("="*70)

print(seasonal_results_df)

seasonal_results_df.to_csv(
    "seasonal_results.csv",
    index=False
)
```
